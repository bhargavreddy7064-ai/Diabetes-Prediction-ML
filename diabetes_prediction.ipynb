Set Up and Download Dataset

```python
import urllib.request

# Download the Pima Indians Diabetes dataset
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
file_name = "diabetes.csv"
urllib.request.urlretrieve(url, file_name)

print("Dataset downloaded successfully!")
```

Load and Inspect Data

```python
import numpy as np
import pandas as pd


column_names = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
    'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome'
]


df = pd.read_csv('diabetes.csv', names=column_names)


print(df.head())
print("\nDataset Shape:", df.shape)
print("\nClass Distribution:\n", df['Outcome'].value_counts())
```

Data Preprocessing & Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Separate features (X) and target variable (y)
X = df.drop(columns='Outcome', axis=1)
y = df['Outcome']

# Standardize feature values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split dataset into 80% Training set and 20% Testing set
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, stratify=y, random_state=42
)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size: {X_test.shape[0]} samples")
```

Train Machine Learning Model

```python
from sklearn.svm import SVC

# Initialize Support Vector Machine Classifier
classifier = SVC(kernel='linear', random_state=42)

# Train model on training data
classifier.fit(X_train, y_train)
print("Model training complete!")
```

Evaluate Model Performance

```python
from sklearn.metrics import accuracy_score, classification_report

# Predictions on training data
y_train_pred = classifier.predict(X_train)
train_acc = accuracy_score(y_train, y_train_pred)

# Predictions on testing data
y_test_pred = classifier.predict(X_test)
test_acc = accuracy_score(y_test, y_test_pred)

print(f"Training Accuracy: {train_acc * 100:.2f}%")
print(f"Testing Accuracy: {test_acc * 100:.2f}%")
print("\nClassification Report:\n", classification_report(y_test, y_test_pred))
```

Build Predictive System


```python
# Raw input data for a patient:
# [Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin, BMI, DPF, Age]
input_patient_data = (5, 166, 72, 19, 175, 25.8, 0.587, 51)

# Convert tuple to numpy array and reshape for single instance prediction
input_array = np.asarray(input_patient_data).reshape(1, -1)

# Standardize input array using previously fitted scaler
std_input = scaler.transform(input_array)

# Predict outcome
prediction = classifier.predict(std_input)

if prediction[0] == 1:
    print("Prediction: The person is likely Diabetic.")
else:
    print("Prediction: The person is Non-Diabetic.")
```
